```python
print(X["Sex"].value_counts())   
print(X.isnull().sum())          
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)
X_smote, y_smote = smote.fit_resample(X, y)

print(y_smote.value_counts())
```

```python
print(X.shape)         
print(X_smote.shape)   
```

```python
import pandas as pd

# séparer les deux classes
classe_0 = df_standarisation[df_standarisation["Survived"] == 0]   
classe_1 = df_standarisation[df_standarisation["Survived"] == 1]   

# garder au hasard 342 lignes de la classe 0
classe_0_reduite = classe_0.sample(n=len(classe_1), random_state=42)

# recoller les deux classes
df_under = pd.concat([classe_0_reduite, classe_1])

print(df_under["Survived"].value_counts())   # 342 et 342
print(df_under.shape)                        # (684, 12)
```

```python
X_under = df_under.drop(columns=["Survived", "PassengerId", "Name", "Ticket"])
y_under = df_under["Survived"]
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(15, 4))

plt.subplot(1, 3, 1)
sns.countplot(x=y)
plt.title("Original")

plt.subplot(1, 3, 2)
sns.countplot(x=df_under["Survived"])
plt.title("Undersampling")

plt.subplot(1, 3, 3)
sns.countplot(x=y_smote)
plt.title("SMOTE")

plt.show()
```
